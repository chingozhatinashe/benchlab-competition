the submitted config fitted on slow runs) when it is high. Pilot fold 2024 only; aborted.
Study: `python -m training.pfss_switch`; write-up `docs/pfss_switch.md`.

```python
import pandas as pd

from training import pfss_switch as ps

d = pd.read_csv(ps.EXPERIMENTS / "pfss_switch.csv").set_index("variant")
ref = d.loc["reference"]
out = pd.DataFrame({
    "vs reference %": 100 * (d["all"] / ref["all"] - 1),
    "slow runs %": 100 * (d["slow wind"] / ref["slow wind"] - 1),
    "stream arrival %": 100 * (d["stream arrival"] / ref["stream arrival"] - 1),
    "CME %": 100 * (d["CME"] / ref["CME"] - 1),
})
out.sort_values("vs reference %").round(2)
```
