# Post-hoc corrections of the linear forecast (not built)

A persistence blend and calibration models of the residual, fitted on the year before the fold
(nested) or on the fold's own fit runs (in-sample), applied to the submitted linear model.
Run on the 2024 fold only: no variant came near the rule, so the other folds were not run.
Study: `python -m training.post_calibration`; write-up `docs/post_calibration.md`.

```python
import pandas as pd

from training import post_calibration as pc

d = pd.read_csv(pc.EXPERIMENTS / "post_calibration.csv").set_index("variant")
ref = d.loc["reference"]
out = pd.DataFrame({
    "vs reference %": 100 * (d["all"] / ref["all"] - 1),
    "slow runs vs reference %": 100 * (d["slow wind"] / ref["slow wind"] - 1),
    "CME vs reference %": 100 * (d["CME"] / ref["CME"] - 1),
    "slow residual": d["residual slow wind"],
    "all residual": d["residual all"],
})
out.sort_values("vs reference %").round(2)
```

Every correction is worse than the reference. Fitted on the year before, they move the
forecast up (the fold's mean residual goes from −52 to about −62 km/s): with the official
weights, which stress the under-forecast fast hours, the weighted residual of the year before
is +10 km/s although its plain mean is −23 km/s. Fitted in-sample, there is almost nothing
left to correct.
